```python
# Convert Order Date to datetime
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    errors='coerce'
)

# Extract month
df['Month'] = df['Order Date'].dt.month

# Calculate average sales by month
seasonal_sales = (
    df.groupby('Month')['Total Amount']
      .mean()
      .sort_index()
)

print(seasonal_sales)
```

```python
highest_month = seasonal_sales.idxmax()
highest_sales = seasonal_sales.max()

lowest_month = seasonal_sales.idxmin()
lowest_sales = seasonal_sales.min()

print(f"Highest Sales Month: {highest_month}")
print(f"Highest Average Sales: ₹{highest_sales:.2f}")

print(f"Lowest Sales Month: {lowest_month}")
print(f"Lowest Average Sales: ₹{lowest_sales:.2f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Convert Order Date to datetime
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    errors='coerce'
)

# 2. Calculate monthly revenue by category
monthly_category_revenue = (
    df.groupby([
        pd.Grouper(key='Order Date', freq='M'),
        'Category'
    ])['Total Amount']
    .sum()
    .unstack()
)

# 3. Display the data
print(monthly_category_revenue)

# 4. Create line chart
monthly_category_revenue.plot(
    kind='line',
    marker='o',
    figsize=(12, 6)
)

# 5. Add labels
plt.title('Revenue Trend by Category Across Months')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.legend(title='Category')
plt.grid(True)
plt.tight_layout()

# 6. Display chart
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Convert Order Date to datetime
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    errors='coerce'
)

# 2. Calculate total sales month-wise
monthly_sales = (
    df.resample('M', on='Order Date')['Total Amount']
      .sum()
)

# 3. Display monthly sales
print(monthly_sales)

# 4. Create line chart
monthly_sales.plot(
    kind='line',
    marker='o',
    figsize=(10, 5)
)

# 5. Add chart details
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.tight_layout()

# 6. Display chart
plt.show()
```

```python
# 1. Convert Order Date to datetime
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    errors='coerce'
)

# 2. Calculate monthly revenue for each category
monthly_category_revenue = (
    df.groupby([
        pd.Grouper(key='Order Date', freq='M'),
        'Category'
    ])['Total Amount']
    .sum()
    .unstack()
)

# 3. Display the calculated data
print(monthly_category_revenue)

# 4. Create line chart
monthly_category_revenue.plot(
    kind='line',
    marker='o',
    figsize=(12, 6)
)

# 5. Add chart details
plt.title('Revenue Trend by Category Across Months')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.legend(title='Category')
plt.grid(True)
plt.tight_layout()

# 6. Display chart
plt.show()
```

```python
# 1. Calculate sales by payment mode
payment_sales = (
    df.groupby('Payment Mode')['Total Amount']
      .sum()
      .sort_values(ascending=False)
)

# 2. Display the values
print(payment_sales)

# 3. Create pie chart
plt.figure(figsize=(8, 8))

plt.pie(
    payment_sales,
    labels=payment_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

# 4. Add title
plt.title('Sales Contribution by Payment Mode')

# 5. Keep pie chart circular
plt.axis('equal')

# 6. Display chart
plt.show()
```

```python
# 1. Count unique orders by category
category_orders = (
    df.groupby('Category')['Order ID']
      .nunique()
      .sort_values(ascending=False)
)

# 2. Display order counts
print(category_orders)

# 3. Create pie chart
plt.figure(figsize=(8, 8))

plt.pie(
    category_orders,
    labels=category_orders.index,
    autopct='%1.1f%%',
    startangle=90
)

# 4. Add title
plt.title('Share of Orders by Category')

# 5. Keep pie chart circular
plt.axis('equal')

# 6. Display chart
plt.show()
```
